##### Questão 3: Explique, do ponto de vista teórico:

* o que são variáveis dummies;
    - Variáveis dummies (ou variáveis binárias) são variáveis qualitativas representadas numericamente pelos valores 0 e 1, indicando, respectivamente, a ausência ou a presença de uma determinada característica. Elas permitem incorporar informações categóricas em modelos econométricos, como sexo, estado civil, escolaridade, região geográfica ou setor de atividade.

* qual sua finalidade em modelos econométricos; 
    - A principal finalidade das variáveis dummies é permitir que características qualitativas influenciem a variável dependente em um modelo econométrico. Elas possibilitam estimar o efeito de atributos que não podem ser medidos numericamente, como gênero, região, ocupação ou nível de escolaridade.
    Em modelos de regressão, os coeficientes associados às variáveis dummies representam a diferença média entre um grupo específico e a categoria de referência, mantendo constantes as demais variáveis explicativas. Dessa forma, é possível avaliar, por exemplo, se indivíduos com ensino superior apresentam renda diferente daqueles sem ensino superior, controlando outros fatores do modelo.

* o significado da categoria de referência;
    - A categoria de referência (ou categoria-base) é o grupo utilizado como comparação para interpretar os coeficientes das variáveis dummies. Essa categoria não recebe uma variável dummy própria quando o modelo possui intercepto, evitando problemas de multicolinearidade.
     - Por exemplo, considerando a variável gênero:
        - Mulher = 1
        - Homem = 0

    - Os homens constituem a categoria de referência. Assim, o coeficiente estimado para a dummy "Mulher" representa a diferença média entre mulheres e homens, mantendo constantes as demais variáveis do modelo.

* o problema da armadilha das variáveis dummies (Dummy Variable Trap);
   - A Dummy Variable Trap ocorre quando todas as categorias de uma variável qualitativa são incluídas no modelo juntamente com o intercepto. Nessa situação, uma dummy pode ser escrita como combinação linear das demais, gerando multicolinearidade perfeita, o que impede a estimação dos coeficientes pelo método dos Mínimos Quadrados Ordinários (MQO).
    Por exemplo, para representar quatro regiões (Norte, Nordeste, Sul e Sudeste), não se deve incluir quatro variáveis dummies juntamente com o intercepto, pois a soma das quatro será sempre igual a 1. A solução consiste em: 
        - manter o intercepto e utilizar apenas k − 1 variáveis dummies, deixando uma categoria como referência; ou incluir todas as k dummies e retirar o intercepto do modelo.
        - Na prática, a abordagem mais utilizada é manter o intercepto e omitir uma das categorias, permitindo que todos os coeficientes sejam interpretados em relação à categoria de referência.

##### Questão 5: Estime inicialmente um Modelo de Probabilidade Linear (MQO)

```python
#a. equação do modelo de regressão linear múltipla (MPL) para a variável dependente "income" e as variáveis independentes "age", "educational-num", "hours-per-week", "gender" e "occupation".
import statsmodels.formula.api as smf

modelo_mpl = smf.ols(
    'income ~ age + Q("educational-num") + Q("hours-per-week") + C(gender) + C(occupation)',
    data=df
).fit()

print(modelo_mpl.summary())
modelo_mpl.summary()
```

Como esse valor é negativo, ele não possui interpretação econômica direta, evidenciando uma limitação do Modelo de Probabilidade Linear.
    * Male (0,1566): indivíduos do sexo masculino possuem uma probabilidade aproximadamente 15,66 pontos percentuais maior de apresentar renda superior a US$ 50.000 em relação às mulheres, mantendo as demais variáveis constantes.
    * Age (0,0056): cada ano adicional de idade aumenta em aproximadamente 0,56 ponto percentual a probabilidade de um indivíduo possuir renda superior a US$ 50.000.
    * Educational-num (0,0356): cada ano adicional de escolaridade aumenta em aproximadamente 3,56 pontos percentuais a probabilidade de o indivíduo possuir renda superior a US$ 50.000.

#### d. limitações do modelo:


* Segundo Gujarati, o Modelo de Probabilidade Linear apresenta algumas limitações importantes:

    * As probabilidades estimadas podem assumir valores menores que 0 ou maiores que 1, o que não faz sentido do ponto de vista probabilístico.
    * Os erros tendem a apresentar heterocedasticidade, violando uma das hipóteses do Modelo Clássico de Regressão Linear.
    * O efeito marginal das variáveis explicativas é constante, independentemente do nível da probabilidade, o que pode não representar adequadamente o comportamento dos dados.

##### Questão 6: Explique teoricamente os modelos Logit e Probit. Responda:

* a. Por que o MQO não é o método mais adequado para variável dependente binária?
    - O Modelo de Probabilidade Linear (MQO) apresenta algumas limitações quando a variável dependente é binária:
        - pode gerar probabilidades menores que 0 ou maiores que 1;
        - os erros tendem a apresentar heterocedasticidade, violando uma hipótese do Modelo Clássico de Regressão Linear;
        - assume que o efeito das variáveis explicativas sobre a probabilidade é constante, o que nem sempre representa adequadamente a realidade.

* b. Diferenças entre os modelos Logit e Probit
    - A principal diferença entre os modelos está na distribuição utilizada para representar a probabilidade.
        - Modelo Logit: utiliza a distribuição logística para modelar a probabilidade do evento ocorrer.
        - Modelo Probit: utiliza a distribuição normal padrão (distribuição acumulada da normal).

* c. Vantagens e limitações de cada modelo
    - Modelo Logit
        - Vantagens:
            - garante probabilidades entre 0 e 1;
            - possui interpretação baseada na razão de chances (odds ratio);
            - é amplamente utilizado em estudos econômicos, financeiros e sociais.

        - Limitações:
            - os coeficientes não representam diretamente o efeito sobre a probabilidade, sendo necessária a análise dos efeitos marginais para uma interpretação mais intuitiva.
    - Modelo Probit
        - Vantagens:
            - garante probabilidades entre 0 e 1;
            - baseia-se na distribuição normal, adequada quando se assume que existe uma variável latente normalmente distribuída;
            - também produz estimativas consistentes para variáveis binárias.
        - Limitações:
            - os coeficientes também não possuem interpretação direta em termos de probabilidade;
            - interpretação costuma ser menos intuitiva do que a do modelo Logit.

##### Questão 8: Estime um modelo Probit

```python
import statsmodels.formula.api as smf

modelo_probit = smf.probit(
    'income ~ age + Q("educational-num") + Q("hours-per-week") + C(gender) + C(occupation)',
    data=df
).fit()

print(modelo_probit.summary())
```

a. Equação estimada
-A equação estimada do modelo Probit é:
    - Φ**−1(P(income=1))= −5,0926+0,6472(Male)+0,3854(Adm-clerical)+0,7184(Armed-Forces)+0,4460(Craft-repair)0,8656(Exec-managerial)−0,2160(Farming-fishing)−0,0533(Handlers-cleaners)+0,2475(Machine-op-inspct)−0,1679(Other-service)+⋯+0,0260(Age)+0,1548(Educational-num)+0,0183(Hours-per-week)

b. Interpretação dos coeficientes
- No modelo Probit, os coeficientes não representam diretamente a variação da probabilidade. Eles indicam como cada variável influencia um índice latente que determina a probabilidade de o indivíduo possuir renda superior a US$ 50.000.
    - Coeficiente positivo: aumenta a probabilidade de income = 1.
    - Coeficiente negativo: reduz a probabilidade de income = 1.

    - O coeficiente de Male (0,6472) é positivo, indicando que homens apresentam maior probabilidade de possuir renda superior a US$ 50.000 quando comparados às mulheres.
    - A variável Age (0,0260) possui efeito positivo, indicando que indivíduos mais velhos tendem a apresentar maior probabilidade de renda elevada.
    - Educational-num (0,1548) apresenta coeficiente positivo, sugerindo que maior escolaridade aumenta a probabilidade de renda superior a US$ 50.000.
    - Hours-per-week (0,0183) também apresenta efeito positivo, indicando que mais horas trabalhadas por semana aumentam essa probabilidade.
    - Entre as ocupações, Exec-managerial, Craft-repair, Adm-clerical e Machine-op-inspct possuem efeito positivo, enquanto Farming-fishing e Other-service apresentam efeito negativo em relação à categoria de referência.

c. Comparação com o modelo Logit
- Os resultados do modelo Probit foram bastante semelhantes aos obtidos pelo modelo Logit.
- Ambos apresentaram os mesmos sinais para os coeficientes, indicando a mesma direção dos efeitos das variáveis explicativas.
- As variáveis idade, escolaridade, horas trabalhadas e gênero permaneceram estatisticamente significativas nos dois modelos.
- As ocupações Exec-managerial, Craft-repair e Adm-clerical apresentaram efeito positivo e significativo nos dois modelos, enquanto Farming-fishing e Other-service apresentaram efeito negativo.
- Algumas categorias, como Handlers-cleaners, não foram estatisticamente significativas em ambos os modelos.
- A categoria Armed-Forces foi significativa no Logit (valor-p = 0,031), mas no Probit apresentou valor-p = 0,054, ficando ligeiramente acima do nível de significância de 5%. Assim, no Probit ela não é considerada estatisticamente significativa a 5%, embora seja significativa a 10%.

- Os efeitos marginais convertem esses coeficientes em mudanças na probabilidade, tornando a interpretação econômica mais simples e intuitiva.
