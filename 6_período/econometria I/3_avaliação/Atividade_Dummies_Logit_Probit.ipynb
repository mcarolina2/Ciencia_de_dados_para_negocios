# Atividade - Variáveis Dummies, MQO, Logit e Probit

## 1. Bibliotecas

```
import pandas as pd
import statsmodels.formula.api as smf
from sklearn.metrics import confusion_matrix, accuracy_score
```

## 2. Base de dados

```
df=pd.read_csv('adult.csv')
df['income']=df['income'].str.strip().map({'<=50K':0,'>50K':1})
```

## 3. MQO

```
modelo_mqo=smf.ols('income ~ age + Q("educational-num") + Q("hours-per-week") + C(gender)+C(occupation)',data=df).fit()
print(modelo_mqo.summary())
```

## 4. Logit

```
modelo_logit=smf.logit('income ~ age + Q("educational-num") + Q("hours-per-week") + C(gender)+C(occupation)',data=df).fit()
print(modelo_logit.summary())
```

## 5. Probit

```
modelo_probit=smf.probit('income ~ age + Q("educational-num") + Q("hours-per-week") + C(gender)+C(occupation)',data=df).fit()
print(modelo_probit.summary())
```

## 6. Efeitos marginais

```
print(modelo_logit.get_margeff().summary())
```

## 7. Matriz de confusão

```
pred=(modelo_logit.predict(df)>=0.5).astype(int)
cm=confusion_matrix(df['income'],pred)
print(cm)
print(accuracy_score(df['income'],pred))
```
