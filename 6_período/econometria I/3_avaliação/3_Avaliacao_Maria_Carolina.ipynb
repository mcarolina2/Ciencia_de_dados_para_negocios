##### Questão 3: Explique, do ponto de vista teórico:

* o que são variáveis dummies;
    - Variáveis dummies (ou variáveis binárias) são variáveis qualitativas representadas numericamente pelos valores 0 e 1, indicando, respectivamente, a ausência ou a presença de uma determinada característica. Elas permitem incorporar informações categóricas em modelos econométricos, como sexo, estado civil, escolaridade, região geográfica ou setor de atividade.

* qual sua finalidade em modelos econométricos; 
    -A principal finalidade das variáveis dummies é permitir que características qualitativas influenciem a variável dependente em um modelo econométrico. Elas possibilitam estimar o efeito de atributos que não podem ser medidos numericamente, como gênero, região, ocupação ou nível de escolaridade.
    Em modelos de regressão, os coeficientes associados às variáveis dummies representam a diferença média entre um grupo específico e a categoria de referência, mantendo constantes as demais variáveis explicativas. Dessa forma, é possível avaliar, por exemplo, se indivíduos com ensino superior apresentam renda diferente daqueles sem ensino superior, controlando outros fatores do modelo.

* o significado da categoria de referência;
    - A categoria de referência (ou categoria-base) é o grupo utilizado como comparação para interpretar os coeficientes das variáveis dummies. Essa categoria não recebe uma variável dummy própria quando o modelo possui intercepto, evitando problemas de multicolinearidade.
     - Por exemplo, considerando a variável gênero:
        - Mulher = 1
        - Homem = 0

    - Os homens constituem a categoria de referência. Assim, o coeficiente estimado para a dummy "Mulher" representa a diferença média entre mulheres e homens, mantendo constantes as demais variáveis do modelo.

* o problema da armadilha das variáveis dummies (Dummy Variable Trap);
   - A Dummy Variable Trap ocorre quando todas as categorias de uma variável qualitativa são incluídas no modelo juntamente com o intercepto. Nessa situação, uma dummy pode ser escrita como combinação linear das demais, gerando multicolinearidade perfeita, o que impede a estimação dos coeficientes pelo método dos Mínimos Quadrados Ordinários (MQO).
    Por exemplo, para representar quatro regiões (Norte, Nordeste, Sul e Sudeste), não se deve incluir quatro variáveis dummies juntamente com o intercepto, pois a soma das quatro será sempre igual a 1. A solução consiste em: 
        - manter o intercepto e utilizar apenas k − 1 variáveis dummies, deixando uma categoria como referência; ou incluir todas as k dummies e retirar o intercepto do modelo.
        - Na prática, a abordagem mais utilizada é manter o intercepto e omitir uma das categorias, permitindo que todos os coeficientes sejam interpretados em relação à categoria de referência.
